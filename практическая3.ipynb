Задание 10.

```python
genre_sales.sort_values(ascending=False).plot(kind="bar", figsize=(10, 5))
plt.title("Мировые продажи по жанрам")
plt.ylabel("Продажи,млн")
plt.xlabel("Жанр")
plt.xticks(rotation=45)
plt.show()
```

```python
# 1. Наибольшие продажи имеет жанр Action
# 2. Наименьшие продажи имеет жанр Strategy
# 3. Разница между максимальным и минимальным значением составляет около 1500
```

Задание 12.

```python
plt.figure(figsize=(8, 5))
plt.scatter(
    df_clean["NA_Sales"],
    df_clean["Global_Sales"],
    alpha=0.5
)
plt.title("Связь продаж в Северной Америке и мировых продаж")
plt.ylabel("Продажи, млн")
plt.xlabel("Продажи в Северной Америке, млн")
plt.xticks(rotation=45)
plt.show()
```

```python
# 1. Одна точка - одна игра.
# 2. Да, наблюдается.
# 3. Соответствует коэффициенту корреляции ≈0.94.
```
